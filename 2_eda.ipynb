```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load Cleaned Data & Set Style
df = pd.read_csv('../data/cleaned_data.csv')
sns.set_theme(style="whitegrid")
colors = ["#2ecc71", "#e74c3c"] # Green for No Churn, Red for Churn

# Create a figure with subplots for a professional layout
fig = plt.figure(figsize=(18, 22))
fig.suptitle('Customer Churn Analysis - Exploratory Data Analysis', fontsize=24, fontweight='bold', y=0.98)

# Visualization 1: Overall Churn Distribution (Pie Chart)
ax1 = plt.subplot(3, 2, 1)
churn_counts = df['Churn_Label'].value_counts()
ax1.pie(churn_counts, labels=churn_counts.index, autopct='%1.1f%%', colors=colors, startangle=90, explode=(0, 0.1))
ax1.set_title('1. Overall Customer Churn Rate', fontsize=16, fontweight='bold')

# Visualization 2: Churn by Contract Type (Bar Chart)
ax2 = plt.subplot(3, 2, 2)
sns.countplot(data=df, x='Contract', hue='Churn_Label', palette=colors, ax=ax2)
ax2.set_title('2. Churn by Contract Type', fontsize=16, fontweight='bold')
ax2.set_xlabel('Contract Type', fontsize=12)
ax2.set_ylabel('Number of Customers', fontsize=12)

# Visualization 3: Churn by Tenure Group (Bar Chart)
ax3 = plt.subplot(3, 2, 3)
tenure_order = ['0-1 Year', '1-2 Years', '2-3 Years', '3-4 Years', '4-5 Years', '5+ Years']
sns.countplot(data=df, x='Tenure_Group', hue='Churn_Label', order=tenure_order, palette=colors, ax=ax3)
ax3.set_title('3. Churn vs. Customer Tenure', fontsize=16, fontweight='bold')
ax3.set_xlabel('Tenure Group', fontsize=12)
ax3.set_ylabel('Number of Customers', fontsize=12)
plt.xticks(rotation=45)

# Visualization 4: Monthly Charges vs Churn (Boxplot)
ax4 = plt.subplot(3, 2, 4)
sns.boxplot(data=df, x='Churn_Label', y='MonthlyCharges', palette=colors, ax=ax4)
ax4.set_title('4. Monthly Charges vs. Churn', fontsize=16, fontweight='bold')
ax4.set_xlabel('Churned?', fontsize=12)
ax4.set_ylabel('Monthly Charges ($)', fontsize=12)

# Visualization 5: Correlation Heatmap (Advanced Analysis prep)
ax5 = plt.subplot(3, 1, 3)
numeric_df = df[['Tenure', 'MonthlyCharges', 'TotalCharges', 'SeniorCitizen', 'Churn']]
correlation = numeric_df.corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5, ax=ax5)
ax5.set_title('5. Correlation Heatmap of Numeric Variables', fontsize=16, fontweight='bold')

plt.tight_layout(pad=3.0)
plt.show()
```
